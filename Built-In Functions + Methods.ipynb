```python
# Demonstrating Python built-in functions

# abs() - returns the absolute value of a number
print(abs(-7.5))  # Output: 7.5

# all() - returns True if all elements in an iterable are true
print(all([True, True, False]))  # Output: False

# any() - returns True if any element in an iterable is true
print(any([False, False, True]))  # Output: True

# bin() - converts an integer to a binary string
print(bin(10))  # Output: '0b1010'

# bool() - converts a value to a Boolean
print(bool(0))  # Output: False

# chr() - returns the character that represents the specified Unicode code
print(chr(97))  # Output: 'a'

# divmod() - returns a tuple containing the quotient and remainder when dividing two numbers
print(divmod(10, 3))  # Output: (3, 1)

# enumerate() - adds a counter to an iterable and returns it as an enumerate object
for index, value in enumerate(['a', 'b', 'c']):
    print(index, value)  # Output: 0 a, 1 b, 2 c

# filter() - constructs an iterator from elements of an iterable for which a function returns true
print(list(filter(lambda x: x > 0, [-1, 0, 1, 2])))  # Output: [1, 2]

# float() - converts a value to a floating-point number
print(float('3.14'))  # Output: 3.14

# int() - converts a value to an integer
print(int('10'))  # Output: 10

# len() - returns the length of an object
print(len('hello'))  # Output: 5

# max() - returns the largest item in an iterable or the largest of two or more arguments
print(max(1, 2, 3))  # Output: 3

# min() - returns the smallest item in an iterable or the smallest of two or more arguments
print(min(1, 2, 3))  # Output: 1

# pow() - returns the value of x to the power of y
print(pow(2, 3))  # Output: 8

# range() - returns a sequence of numbers, starting from 0 by default, and increments by 1 by default
print(list(range(5)))  # Output: [0, 1, 2, 3, 4]

# round() - rounds a number to a specified number of decimal places
print(round(3.14159, 2))  # Output: 3.14

# sorted() - returns a sorted list of the specified iterable
print(sorted([3, 1, 2]))  # Output: [1, 2, 3]

# sum() - returns the sum of all items in an iterable
print(sum([1, 2, 3]))  # Output: 6

# type() - returns the type of an object
print(type(123))  # Output: <class 'int'>

# zip() - returns an iterator of tuples, where the i-th tuple contains the i-th element from each of the argument iterables
print(list(zip([1, 2, 3], ['a', 'b', 'c'])))  # Output: [(1, 'a'), (2, 'b'), (3, 'c')]
```
