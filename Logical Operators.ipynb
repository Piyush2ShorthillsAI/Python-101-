```python
# Define variables
x = 3

# and operator
result_and = (x < 5 and x < 10)
print(f"x < 5 and x < 10: {result_and}")

# or operator
result_or = (x < 5 or x < 4)
print(f"x < 5 or x < 4: {result_or}")

# not operator
result_not = not(x < 5 and x < 10)
print(f"not(x < 5 and x < 10): {result_not}")
```
