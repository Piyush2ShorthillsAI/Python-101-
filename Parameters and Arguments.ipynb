```python
# Function with one argument
def my_function(fname):
    print(fname + " Refsnes")

my_function("Emil")
my_function("Tobias")
my_function("Linus")

# Function with two arguments
def my_function(fname, lname):
    print(fname + " " + lname)

my_function("Emil", "Refsnes")

# Uncommenting the following line will cause an error because the function expects 2 arguments but gets only 1
# my_function("Emil")
```
