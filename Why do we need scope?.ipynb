```python
# Scopes are essential in programming because they 
# define the visibility and lifetime of variables.
#  By organizing variables into different scopes, 
#  we can prevent naming conflicts and ensure that 
#  variables are only accessible where they are 
#  intended to be used. This helps in maintaining
#   modularity and readability of the code. For i
#   nstance, variables defined within a function 
#   are not accessible outside of it, which prevents 
#   unintended interactions with other parts of the p
# rogram. Scopes also enable the use of the same 
# variable names in different parts of a program 
# without causing conflicts, as each scope is isolated. 
# This encapsulation of variables within their respective 
# scopes is crucial for debugging and maintaining large 
# codebases, as it reduces the risk of side effects and 
# makes the 
# code more predictable and easier to understand.

# Global scope
x = 10
y = 50

def outer_function():
    # Enclosing scope
    x = 20
    y = 60
    
    def inner_function():
        # Local scope
        x = 30
        y = 70
        print("Inner function x:", x)  # Should print 30
        print("Inner function y:", y)  # Should print 70
    
    inner_function()
    print("Outer function x:", x)  # Should print 20
    print("Outer function y:", y)  # Should print 60

outer_function()
print("Global scope x:", x)  # Should print 10
print("Global scope y:", y)  # Should print 50

# Another example to show variable shadowing and scope
def another_outer_function():
    # Enclosing scope
    a = 100
    
    def another_inner_function():
        # Local scope
        nonlocal a
        a = 200
        print("Another inner function a:", a)  # Should print 200
    
    another_inner_function()
    print("Another outer function a:", a)  # Should print 200

another_outer_function()

# Global scope variable
a = 300
print("Global scope a:", a)  # Should print 300
```
