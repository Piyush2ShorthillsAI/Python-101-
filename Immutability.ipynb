```python
# Immutable Objects in Python

# Example 1: Tuples are immutable
try:
    tuple1 = (0, 1, 2, 3)
    tuple1[0] = 4
    print(tuple1)
except TypeError as e:
    print(f"Error: {e}")

# Example 2: Strings are immutable
try:
    message = "Welcome to GeeksforGeeks"
    message[0] = 'p'
    print(message)
except TypeError as e:
    print(f"Error: {e}")

# Mutable Objects in Python

# Example 1: Lists are mutable
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)

my_list.insert(1, 5)
print(my_list)

my_list.remove(2)
print(my_list)

popped_element = my_list.pop(0)
print(my_list)
print(popped_element)

# Example 2: Dictionaries are mutable
my_dict = {"name": "Ram", "age": 25}
new_dict = my_dict
new_dict["age"] = 37

print(my_dict)
print(new_dict)

# Example 3: Sets are mutable
my_set = {1, 2, 3}
new_set = my_set
new_set.add(4)

print(my_set)
print(new_set)

# Exception in immutability
tup = ([3, 4, 5], 'myname')
tup[0][0] = 10  # Modifying the list inside the tuple
print(tup)
```
