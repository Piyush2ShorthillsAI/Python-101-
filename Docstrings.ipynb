```python
# A docstring is a special type of comment used to describe the purpose of a function or a class.
# It is written inside triple quotes (""" or ''') and can span multiple lines.
# Docstrings are used to provide documentation for your code and can be accessed using the help() function or __doc__ attribute.

def add_numbers(a, b):
    """
    Add two numbers and return the result.

    Parameters:
    a (int or float): The first number.
    b (int or float): The second number.

    Returns:
    int or float: The sum of the two numbers.

    Example:
    >>> add_numbers(2, 3)
    5
    >>> add_numbers(2.5, 3.5)
    6.0
    """
    return a + b

# Example usage
result = add_numbers(2, 3)
print(result)  # Output: 5
```
