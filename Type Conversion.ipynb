```python
# Integer to float
int_value = 10
float_value = float(int_value)
print(f"Integer to float: {float_value}")

# Float to integer
float_value = 10.5
int_value = int(float_value)
print(f"Float to integer: {int_value}")

# Integer to string
int_value = 10
str_value = str(int_value)
print(f"Integer to string: {str_value}")

# String to integer
str_value = "10"
int_value = int(str_value)
print(f"String to integer: {int_value}")

# String to float
str_value = "10.5"
float_value = float(str_value)
print(f"String to float: {float_value}")

# Float to string
float_value = 10.5
str_value = str(float_value)
print(f"Float to string: {str_value}")

# List to tuple
list_value = [1, 2, 3]
tuple_value = tuple(list_value)
print(f"List to tuple: {tuple_value}")

# Tuple to list
tuple_value = (1, 2, 3)
list_value = list(tuple_value)
print(f"Tuple to list: {list_value}")
```
