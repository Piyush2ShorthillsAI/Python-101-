```python
from math import pi

# Example 1: Local Scope
pi = 'outer pi variable'

def print_pi():
    pi = 'inner pi variable'
    print(pi)

print_pi()
print(pi)

# Example 2: Local Scope
pi = 'global pi variable'

def inner():
    pi = 'inner pi variable'
    print(pi)

inner()

# Example 3: Local and Global Scopes
pi = 'global pi variable'

def inner():
    pi = 'inner pi variable'
    print(pi)

inner()
print(pi)

# Example 4: Local, Enclosed, and Global Scopes
pi = 'global pi variable'

def outer():
    pi = 'outer pi variable'
    def inner():
        # pi = 'inner pi variable'
        nonlocal pi
        print(pi)
    inner()

outer()
print(pi)

# Example 5: Local, Enclosed, Global, and Built-in Scopes

# pi = 'global pi variable'

def outer():
    # pi = 'outer pi variable'
    def inner():
        # pi = 'inner pi variable'
        print(pi)
    inner()

outer()
```
