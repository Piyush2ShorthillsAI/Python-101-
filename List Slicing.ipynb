```python
# Example: Get the items from a list starting at position 1 and ending at position 4 (exclusive).
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Get elements from index 1 to 4 (excluded)
print(a[1:4])

# Output
# [2, 3, 4]

# Get all the items from a list
# Get all elements in the list
print(a[::])
print(a[:])

# Output
# [1, 2, 3, 4, 5, 6, 7, 8, 9]
# [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Get all items before/after a specific position
# Get elements starting from index 2 to the end of the list
b = a[2:]
print(b)

# Get elements starting from index 0 to index 3 (excluding 3rd index)
c = a[:3]
print(c)

# Output
# [3, 4, 5, 6, 7, 8, 9]
# [1, 2, 3]

# Get all items between two positions
# Get elements from index 1 to index 4 (excluding index 4)
b = a[1:4]
print(b)

# Output
# [2, 3, 4]

# Get items at specified intervals
# Get every second element from the list starting from the beginning
b = a[::2]
print(b)

# Get every third element from the list starting from index 2 to 8 (exclusive)
c = a[1:8:3]
print(c)

# Output
# [1, 3, 5, 7, 9]
# [2, 5, 8]

# Out-of-bound slicing
# Example: The slice a[7:15] starts at index 7 and attempts to reach index 15, but since the list ends at index 8, it will return only the available elements (i.e. [8,9]).

# Negative Indexing
# Extract elements using negative indices
# Get elements starting from index -2 to end of list
b = a[-2:]
print(b)

# Get elements starting from index 0 to index -3 (excluding 3rd last index)
c = a[:-3]
print(c)

# Get elements from index -4 to -1 (excluding index -1)
d = a[-4:-1]
print(d)

# Get every 2nd element from index -8 to -1 (excluding index -1)
e = a[-8:-1:2]
print(e)

# Output
# [8, 9]
# [1, 2, 3, 4, 5, 6]
# [6, 7, 8]
# [2, 4, 6, 8]

# Reverse a list using slicing
# Get the entire list using negative step
b = a[::-1]
print(b)

# Output
# [9, 8, 7, 6, 5, 4, 3, 2, 1]
```
