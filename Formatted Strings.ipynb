```python
# Left align
left_aligned = "Left aligned: {:<10}".format(123)
print(left_aligned)

# Right align
right_aligned = "Right aligned: {:>10}".format(123)
print(right_aligned)

# Center align
center_aligned = "Center aligned: {:^10}".format(123)
print(center_aligned)

# Sign to the left most position
sign_left = "Sign left: {:=+10}".format(123)
print(sign_left)

# Plus sign for positive and negative
plus_sign = "Plus sign: {:+10}".format(123)
print(plus_sign)

# Minus sign for negative only
minus_sign = "Minus sign: {:-10}".format(-123)
print(minus_sign)

# Space before positive numbers
space_positive = "Space positive: {: 10}".format(123)
print(space_positive)

# Comma as thousand separator
comma_separator = "Comma separator: {:,}".format(1234567)
print(comma_separator)

# Underscore as thousand separator
underscore_separator = "Underscore separator: {:_}".format(1234567)
print(underscore_separator)

# Binary format
binary_format = "Binary format: {:b}".format(123)
print(binary_format)

# Unicode character
unicode_char = "Unicode character: {:c}".format(123)
print(unicode_char)

# Decimal format
decimal_format = "Decimal format: {:d}".format(123)
print(decimal_format)

# Scientific format (lower case e)
scientific_format_e = "Scientific format (e): {:e}".format(1234567)
print(scientific_format_e)

# Scientific format (upper case E)
scientific_format_E = "Scientific format (E): {:E}".format(1234567)
print(scientific_format_E)

# Fix point number format
fix_point = "Fix point number: {:.2f}".format(123.456)
print(fix_point)

# Fix point number format (uppercase)
fix_point_upper = "Fix point number (upper): {:.2F}".format(123.456)
print(fix_point_upper)

# General format
general_format = "General format: {:g}".format(1234567)
print(general_format)

# General format (uppercase E for scientific)
general_format_upper = "General format (upper): {:G}".format(1234567)
print(general_format_upper)

# Octal format
octal_format = "Octal format: {:o}".format(123)
print(octal_format)

# Hex format (lower case)
hex_format_lower = "Hex format (lower): {:x}".format(123)
print(hex_format_lower)

# Hex format (upper case)
hex_format_upper = "Hex format (upper): {:X}".format(123)
print(hex_format_upper)

# Number format
number_format = "Number format: {:n}".format(1234567)
print(number_format)

# Percentage format
percentage_format = "Percentage format: {:.2%}".format(0.123)
print(percentage_format)
```
