```python
# *args example
def fun(*args):
    return sum(args)

print(fun(1, 2, 3, 4)) 
print(fun(5, 10, 15))   

# **kwargs example
def fun(**kwargs):
    for k, val in kwargs.items():
        print(k, val)

fun(a=1, b=2, c=3)

# Python program to illustrate *args for a variable number of arguments
def myFun(*argv):
    for arg in argv:
        print(arg)

myFun('Hello', 'Welcome', 'to', 'GeeksforGeeks')

# Python program to illustrate *args with a first extra argument
def fun(arg1, *argv):
    print("First argument :", arg1)
    for arg in argv:
        print("Argument *argv :", arg)

fun('Hello', 'Welcome', 'to', 'GeeksforGeeks')

# Python program to illustrate **kwargs for a variable number of keyword arguments
def fun(**kwargs):
    for k, val in kwargs.items():
        print("%s == %s" % (k, val))

# Driver code
fun(s1='Geeks', s2='for', s3='Geeks')

# Python program to illustrate **kwargs with a first extra argument
def fun(arg1, **kwargs):
    for k, val in kwargs.items():
        print("%s == %s" % (k, val))

# Driver code
fun("Hi", s1='Geeks', s2='for', s3='Geeks')

# Using both *args and **kwargs in the same function
def fun(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)

fun(1, 2, 3, a=4, b=5)

# Example function to demonstrate *args and **kwargs
def example_function(*args, **kwargs):
    print(args)    # tuple of positional arguments
    print(kwargs)  # dictionary of keyword arguments

example_function(1, 2, 3, name='Alice', age=30)

# Example function to demonstrate *args is a tuple
def example_function(*args):
    print(type(args))  # <class 'tuple'>

example_function(1, 2, 3)

# Example function to demonstrate the use of *args instead of a list
def sum_values(*args):
    total = sum(args)
    return total

result = sum_values(1, 2, 3, 4, 5)
print(result)  # Output: 15

# Example function to demonstrate how to pass **kwargs
def example_function(**kwargs):
    print(kwargs)

example_function(name='Alice', age=30)
```
