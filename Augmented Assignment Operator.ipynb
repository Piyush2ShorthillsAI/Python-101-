```python
# Addition & Assignment
a = 15
b = 20

a += b
print(a)

# Subtraction & Assignment
a = 107
b = 99

a -= b
print(a)

# Multiplication & Assignment
a = 12
b = 23

a *= b
print(a)

# Division & Assignment
a = 56
b = 5

a /= b
print(a)

# Floor Division & Assignment
a = 56
b = 8

a //= b
print(a)

# Modulo & Assignment
a = 34
b = 5

a %= b
print(a)

# Power & Assignment
a = 5
b = 3

a **= b
print(a)

# Bitwise AND & Assignment
a = 12
b = 10

a &= b
print(a)

# Bitwise OR and Assignment
a = 12
b = 10

a |= b
print(a)

# Bitwise XOR and Assignment
a = 12
b = 10

a ^= b
print(a)

# Bitwise Left Shift and Assignment
a = 17
b = 2

a <<= b
print(a)

# Bitwise Right Shift and Assignment
a = 17
b = 2

a >>= b
print(a)





```
