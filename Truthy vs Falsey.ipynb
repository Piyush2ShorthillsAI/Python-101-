```python
# Example of Truthy and Falsy values in Python

# Initial example with number 7
number = 7
if number: 
    print(number)  # Output: 7

# Changing value of number to 0
number = 0
if number: 
    print(number)  # Output: There is no Output

# Examples of using bool() function to check Truthy and Falsy values
print(bool(7))         # True
print(bool(0))         # False
print(bool([]))        # False
print(bool({7, 4}))    # True
print(bool(-4))        # True
print(bool(0.0))       # False
print(bool(None))      # False
print(bool(1))         # True
print(bool(range(0)))  # False
print(bool(set()))     # False
print(bool([1, 2, 3, 4]))  # True

# Function to check if a number is even or odd
def even_odd(number): 
    if number % 2:  
        # since num % 2 is equal to 1 which is Truthy Value 
        return 'odd number'
    else:  
        # since num % 2 is equal to 0 which is Falsy Value 
        return 'even number'

# Testing the even_odd function
result1 = even_odd(7) 
print(result1)  # Output: odd number

result2 = even_odd(4) 
print(result2)  # Output: even number
```
