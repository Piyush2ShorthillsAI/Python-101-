```python
# Example of unpacking a list in Python

# Define a list of strings
colors = ['red', 'blue', 'green']

# Assign the first, second, and third elements to variables
red, blue, green = colors

print(red)   # Output: red
print(blue)  # Output: blue
print(green) # Output: green

# Example of unpacking with fewer variables (will raise an error)
try:
    red, blue = colors
except ValueError as e:
    print(e)  # Output: ValueError: too many values to unpack (expected 2)

# Example of unpacking and packing
colors = ['red', 'blue', 'green']
red, blue, *other = colors

print(red)   # Output: red
print(blue)  # Output: blue
print(other) # Output: ['green']

# Another example of unpacking and packing
colors = ['cyan', 'magenta', 'yellow', 'black']
cyan, magenta, *other = colors

print(cyan)    # Output: cyan
print(magenta) # Output: magenta
print(other)   # Output: ['yellow', 'black']
```
