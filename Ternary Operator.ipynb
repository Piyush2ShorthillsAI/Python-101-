```python
# Ternary Operator in Python

# Basic Example of Ternary Operator
n = 5
res = "Even" if n % 2 == 0 else "Odd"
print(res)  # Output: Odd

# Ternary Operator in Nested If else
n = -5
res = "Positive" if n > 0 else "Negative" if n < 0 else "Zero"
print(res)  # Output: Negative

# Ternary Operator using Tuple
n = 7
res = ("Odd", "Even")[n % 2 == 0]
print(res)  # Output: Odd

# Ternary Operator using Dictionary
a = 10
b = 20
max_value = {False: a, True: b}[a > b]
print(max_value)  # Output: 20

# Ternary Operator using Python Lambda
a = 10
b = 20
max_value = (lambda x, y: x if x > y else y)(a, b)
print(max_value)  # Output: 20

# Ternary Operator with Print Function
a = 10
b = 20
print("a is greater" if a > b else "b is greater")  # Output: b is greater

# Ternary Operator in Python – FAQs

# What does ternary operator do in Python?
a = 5
b = 10
max_value = a if a > b else b  # max_value will be 10

# What is ternary in dictionary Python?
a = 5
b = 10
my_dict = {'max': a if a > b else b}  # my_dict will be {'max': 10}

# What is a ternary?
# A ternary is a concise way to perform a conditional assignment in a single line.

# How to read ternary operator?
num = -1
result = "Positive" if num > 0 else "Negative"
# Read as: "If num is greater than 0, then result is 'Positive', otherwise result is 'Negative'."

# When to use ternary operator?
a = 5
b = 10
# Instead of this:
if a > b:
    max_value = a
else:
    max_value = b

# Use this:
max_value = a if a > b else b
```
