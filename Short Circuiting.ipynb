```python
# Short-circuiting with 'and'
def check_and(a, b):
    print(f"Evaluating: {a} and {b}")
    return a and b

# Short-circuiting with 'or'
def check_or(a, b):
    print(f"Evaluating: {a} or {b}")
    return a or b

# Test cases
print(check_and(True, False))  # Output: False
print(check_and(True, True))   # Output: True
print(check_and(False, True))  # Output: False (short-circuits, b is not evaluated)
print(check_and(False, False)) # Output: False (short-circuits, b is not evaluated)

print(check_or(False, True))   # Output: True
print(check_or(False, False))  # Output: False
print(check_or(True, False))   # Output: True (short-circuits, b is not evaluated)
print(check_or(True, True))    # Output: True (short-circuits, b is not evaluated)
```
