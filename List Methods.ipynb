```python
# Create a list
my_list = [1, 2, 3, 4, 5]

# Append a single element to the end of the list
my_list.append(6)
print("After append:", my_list)

# Clear all items from the list
my_list.clear()
print("After clear:", my_list)

# Create a new list for further operations
my_list = [1, 2, 3, 4, 5]

# Copy the list
copied_list = my_list.copy()
print("Copied list:", copied_list)

# Count the occurrences of an element in the list
count = my_list.count(3)
print("Count of 3:", count)

# Extend the list by adding elements from an iterable
my_list.extend([6, 7])
print("After extend:", my_list)

# Get the index of an element in the list
index = my_list.index(4)
print("Index of 4:", index)

# Insert an element at a specific position
my_list.insert(2, 10)
print("After insert:", my_list)

# Remove and return an element at a given index
popped_element = my_list.pop(2)
print("Popped element:", popped_element)
print("After pop:", my_list)

# Remove the first occurrence of an element
my_list.remove(4)
print("After remove:", my_list)

# Reverse the list
my_list.reverse()
print("After reverse:", my_list)

# Sort the list
my_list.sort()
print("After sort:", my_list)
```
