```python
import math

# Calculate the square root of a number
sqrt_value = math.sqrt(16)
print("Square root of 16 is" ,sqrt_value)

# Calculate the factorial of a number
factorial_value = math.factorial(5)
print(f"Factorial of 5 is {factorial_value}")

# Calculate the greatest common divisor of two numbers
gcd_value = math.gcd(48, 18)
print(f"GCD of 48 and 18 is {gcd_value}")

# Calculate the sine of an angle in radians
sine_value = math.sin(math.radians(30))
print(f"Sine of 30 degrees is {sine_value}")

# Calculate the exponential of a number
exp_value = math.exp(2)
print(f"Exponential of 2 is {exp_value}")

# Calculate the natural logarithm of a number
log_value = math.log(10)
print(f"Natural logarithm of 10 is {log_value}")

# Calculate the cosine of an angle in radians
cosine_value = math.cos(math.radians(60))
print(f"Cosine of 60 degrees is {cosine_value}")

# Calculate the tangent of an angle in radians
tangent_value = math.tan(math.radians(45))
print(f"Tangent of 45 degrees is {tangent_value}")

# Calculate the hyperbolic sine of a number
sinh_value = math.sinh(1)
print(f"Hyperbolic sine of 1 is {sinh_value}")

# Calculate the hyperbolic cosine of a number
cosh_value = math.cosh(1)
print(f"Hyperbolic cosine of 1 is {cosh_value}")

# Calculate the hyperbolic tangent of a number
tanh_value = math.tanh(1)
print(f"Hyperbolic tangent of 1 is {tanh_value}")

# Calculate the ceiling of a number
ceil_value = math.ceil(4.2)
print(f"Ceiling of 4.2 is {ceil_value}")

# Calculate the floor of a number
floor_value = math.floor(4.8)
print(f"Floor of 4.8 is {floor_value}")

# Calculate the absolute value of a number
abs_value = math.fabs(-5)
print(f"Absolute value of -5 is {abs_value}")

# Calculate the power of a number
pow_value = math.pow(2, 3)
print(f"2 raised to the power of 3 is {pow_value}")

# Calculate the logarithm base 10 of a number
log10_value = math.log10(100)
print(f"Logarithm base 10 of 100 is {log10_value}")
```
