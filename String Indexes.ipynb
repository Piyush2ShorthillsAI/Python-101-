```python
# String Indexing Examples

# Define a string
s = 'mybacon'

# Accessing characters using positive indexing
print(s[0])  # Output: 'm'
print(s[1])  # Output: 'y'
print(s[6])  # Output: 'n'
print(s[len(s) - 1])  # Output: 'n'

# Attempting to access an index out of range
try:
    print(s[7])
except IndexError as e:
    print(e)  # Output: string index out of range

# Accessing characters using negative indexing
print(s[-1])  # Output: 'n'
print(s[-4])  # Output: 'a'
print(s[-len(s)])  # Output: 'm'
print(s[-7])  # Output: 'm'

# Attempting to access an index out of range using negative indexing
try:
    print(s[-8])
except IndexError as e:
    print(e)  # Output: string index out of range

# Working with an empty string
t = ''
print(type(t))  # Output: <class 'str'>

# Attempting to access an index in an empty string
try:
    print(t[0])
except IndexError as e:
    print(e)  # Output: string index out of range

print(len(t))  # Output: 0
```
