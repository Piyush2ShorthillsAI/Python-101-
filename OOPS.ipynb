```python
# Example 1: Objects and Classes
class Computer:
    def config(self):
        print("i5, 16GB RAM, 1TB HDD")

comp1 = Computer()
comp1.config()

# Example 2: The `__init__` Method (Constructor)
class Computer:
    def __init__(self, cpu, ram):
        self.cpu = cpu
        self.ram = ram

    def config(self):
        print(self.cpu, self.ram)

comp1 = Computer("i5", "16GB")
comp1.config()

# Example 3: Types of Methods
class Student:
    school = "XYZ School"  # Class variable

    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def show(self):  # Instance method
        print(self.name, self.age)

    @classmethod
    def get_school(cls):  # Class method
        return cls.school

    @staticmethod
    def info():  # Static method
        print("This is a student class")

s1 = Student("John", 15)
s1.show()
print(Student.get_school())
Student.info()

# Example 4: Comparing Objects
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def compare(self, other):
        return self.age == other.age

p1 = Person("Alice", 30)
p2 = Person("Bob", 30)
print(p1.compare(p2))  # Output: True

# Example 5: Types of Variables
class Car:
    wheels = 4  # Class variable
    
    def __init__(self, brand):
        self.brand = brand  # Instance variable

c1 = Car("BMW")
c2 = Car("Audi")
print(c1.brand, c2.brand, c1.wheels)

# Example 6: Inner Classes
class Student:
    class Laptop:
        def __init__(self, brand, cpu):
            self.brand = brand
            self.cpu = cpu

    def __init__(self, name):
        self.name = name
        self.laptop = self.Laptop("HP", "i5")

s1 = Student("John")
print(s1.laptop.brand)
```
