```python
# Python Dictionary Methods

# Sample dictionary
sample_dict = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}

# clear() - Removes all the elements from the dictionary
sample_dict.clear()
print("After clear():", sample_dict)

# Resetting sample dictionary
sample_dict = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}

# copy() - Returns a copy of the dictionary
copy_dict = sample_dict.copy()
print("Copy of dictionary:", copy_dict)

# fromkeys() - Returns a dictionary with the specified keys and value
keys = ('key1', 'key2', 'key3')
value = 0
new_dict = dict.fromkeys(keys, value)
print("Dictionary from keys:", new_dict)

# get() - Returns the value of the specified key
age = sample_dict.get("age")
print("Value of 'age':", age)

# items() - Returns a list containing a tuple for each key value pair
items = sample_dict.items()
print("Items in dictionary:", items)

# keys() - Returns a list containing the dictionary's keys
keys = sample_dict.keys()
print("Keys in dictionary:", keys)

# pop() - Removes the element with the specified key
removed_value = sample_dict.pop("city")
print("Removed value:", removed_value)
print("Dictionary after pop():", sample_dict)

# popitem() - Removes the last inserted key-value pair
sample_dict["city"] = "New York"  # Re-adding the city key-value pair
last_item = sample_dict.popitem()
print("Last inserted item:", last_item)
print("Dictionary after popitem():", sample_dict)

# setdefault() - Returns the value of the specified key. If the key does not exist: insert the key, with the specified value
default_value = sample_dict.setdefault("country", "USA")
print("Default value:", default_value)
print("Dictionary after setdefault():", sample_dict)

# update() - Updates the dictionary with the specified key-value pairs
sample_dict.update({"age": 30, "city": "Los Angeles"})
print("Dictionary after update():", sample_dict)

# values() - Returns a list of all the values in the dictionary
values = sample_dict.values()
print("Values in dictionary:", values)
```
