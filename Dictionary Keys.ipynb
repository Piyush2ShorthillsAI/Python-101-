```python
# Accessing Items
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Get the value of the "model" key
x = thisdict["model"]
print(x)

# Using get() method to get the value of the "model" key
x = thisdict.get("model")
print(x)

# Get Keys
car = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

x = car.keys()
print(x)  # before the change

car["color"] = "white"
print(x)  # after the change

# Get Values
x = car.values()
print(x)  # before the change

car["year"] = 2020
print(x)  # after the change

car["color"] = "red"
print(x)  # after the change

# Get Items
x = car.items()
print(x)  # before the change

car["year"] = 2020
print(x)  # after the change

car["color"] = "red"
print(x)  # after the change

# Check if Key Exists
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

if "model" in thisdict:
    print("Yes, 'model' is one of the keys in the thisdict dictionary")
```
